```python
routes = [
    {
        "route_id": "R101",
        "departure_city": "Karachi",
        "destination_city": "Hyderabad",
        "departure_time": "08:00 AM",
        "ticket_price": 1200,
        "total_seats": 40,
        "available_seats": 40
    },
    {
        "route_id": "R102",
        "departure_city": "Karachi",
        "destination_city": "Sukkur",
        "departure_time": "09:30 AM",
        "ticket_price": 2000,
        "total_seats": 36,
        "available_seats": 36
    },
    {
        "route_id": "R103",
        "departure_city": "Hyderabad",
        "destination_city": "Mirpur Khas",
        "departure_time": "11:00 AM",
        "ticket_price": 800,
        "total_seats": 32,
        "available_seats": 32
    },
    {
        "route_id": "R104",
        "departure_city": "Karachi",
        "destination_city": "Lahore",
        "departure_time": "07:00 PM",
        "ticket_price": 5000,
        "total_seats": 45,
        "available_seats": 45
    },
    {
        "route_id": "R105",
        "departure_city": "Sukkur",
        "destination_city": "Hyderabad",
        "departure_time": "02:00 PM",
        "ticket_price": 1500,
        "total_seats": 30,
        "available_seats": 30
    }
]

supported_cities = ("Karachi", "Hyderabad", "Sukkur", "Mirpur Khas", "Lahore")
bookings = []
issued_booking_ids = set()

while True:
    print("\nSMART BUS RESERVATION SYSTEM")
    print("1. View Available Routes")
    print("2. Search for a Route")
    print("3. Book a Ticket")
    print("4. View My Bookings")
    print("5. Cancel a Booking")
    print("6. View Seat Availability")
    print("7. Generate Booking Report")
    print("8. Route Schedule")
    print("9. Exit")
    
    choice = int(input("Enter your choice: "))
    
    if choice == 1:
        print("\nAVAILABLE BUS ROUTES")
        print("No. Route\t\t\tDepart\t\tFare\tSeats")
        for index, route in enumerate(routes, start=1):
            status_str = f"{route['available_seats']}/{route['total_seats']}"
            if route['available_seats'] == 0:
                status_str = "Fully Booked"
            print(f"{index}. {route['route_id']} {route['departure_city']} -> {route['destination_city']}\t{route['departure_time']}\tRs.{route['ticket_price']}\t{status_str}")
    elif choice == 2:
        term = input("Search by route ID or city: ").strip().lower()
        found = False
        for route in routes:
            if term == route['route_id'].lower() or term == route['departure_city'].lower() or term == route['destination_city'].lower():
                print(f"{route['route_id']} | {route['departure_city']} -> {route['destination_city']} | {route['departure_time']} | Rs. {route['ticket_price']} | {route['available_seats']} seats available")
                found = True
        if not found:
            print("No matching routes found.")    

    elif choice == 3:
        name = input('Enter the name : ').strip()
        if name == "":
            print("Error : name field is empty!")
            continue
            
        age_input = input('Enter your age : ')
        if not age_input.isdigit():
            print("Error: Age must be a valid number.")
            continue
        age = int(age_input)
        if age <= 0:
            print("Age should be positive")
            continue
            
        route_id = input('Enter your Route_id : ').strip().upper()
        selected_route = None
        for route in routes:
            if route["route_id"] == route_id:
                selected_route = route
                break
                
        if selected_route is None:
            print("Error : ID does not exist")
            continue
            
        seats_input = input('Enter your seats : ')
        if not seats_input.isdigit():
            print("Error: Seat must be an integer.")
            continue
        seats_to_book = int(seats_input)
        
        if seats_to_book <= 0:
            print("Seat must be greater than zero")
            continue
            
        if seats_to_book > selected_route["available_seats"]:
            print("Seat is not available")
            continue
            
        booking_id_num = len(issued_booking_ids) + 1001
        b_id = f"B{booking_id_num}"
        while b_id in issued_booking_ids:
            booking_id_num += 1
            b_id = f"B{booking_id_num}"
            
        issued_booking_ids.add(b_id)
        total_fare = selected_route['ticket_price'] * seats_to_book
        selected_route['available_seats'] -= seats_to_book
        
        new_booking = {
            "booking_id": b_id,
            "passenger_name": name,
            "passenger_age": age,
            "route_id": selected_route['route_id'],
            "seats": seats_to_book,
            "total_fare": total_fare,
            "status": "Confirmed"
        }
        bookings.append(new_booking)
        
        print("\nBooking confirmed!")
        print(f"Booking ID: {b_id}")
        print(f"Passenger: {name}")
        print(f"Route: {selected_route['route_id']} ({selected_route['departure_city']} -> {selected_route['destination_city']})")
        print(f"Seats booked: {seats_to_book}")
        print(f"Fare per seat: Rs. {selected_route['ticket_price']}")
        print(f"Total fare: Rs. {total_fare}")
        print("Status: Confirmed")

    elif choice == 4:
        if bookings ==0:
            print('No Booking Found !')
        else:
            print('My Bookings')
            for index,booking in enumerate(bookings,start=1):
                print(f'''{index}. ID: {booking['booking_id']} | Passenger: {booking['passenger_name']}
                | Route: {booking['route_id']} | Seats: {booking['seats']} | Fare: Rs. {booking['total_fare']} 
                | Status: {booking['status']}''')
    elif choice == 5:
        c_id = input("Enter booking ID to cancel: ").strip().upper()
        found_booking = None
        for booking in bookings:
            if booking['booking_id'] == c_id:
                found_booking = booking
                break
                
        if found_booking is None:
            print(" Booking ID not found.")
        elif found_booking['status'] == "Cancelled":
            print("This booking is already cancelled." )
        else:
            found_booking['status'] = "Cancelled"
            target_route = None
            for route in routes:
                if route['route_id'] == found_booking['route_id']:
                    target_route = route
                    break
            if target_route is not None:
                target_route['available_seats'] += found_booking['seats']
            print(f"Booking {c_id} has been cancelled.")
            print(f"Seats added again: {found_booking['seats']}")
            if target_route is not None:
                print(f"Route {target_route['route_id']} available seats: {target_route['available_seats']}")
    elif choice == 6:
        print("\nSEAT AVAILABILITY")
        for route in routes:
            available = route['available_seats']
            tot = route['total_seats']
            status_text = "Available"
            if available == 0:
                status_text = "Fully Booked"
            elif available < tot:
                status_text = "Partially Booked"
            print(f"Route {route['route_id']} ({route['departure_city']} -> {route['destination_city']}): {available}/{tot} seats available [{status_text}]")
    elif choice == 7:
        total_created = len(bookings)
        confirmed_count = 0
        cancelled_count = 0
        reserved_seats = 0
        confirmed_revenue = 0
        fully_booked_count = 0
        
        for booking in bookings:
            if booking['status'] == "Confirmed":
                confirmed_count += 1
                reserved_seats += booking['seats']
                confirmed_revenue += booking['total_fare']
            elif booking['status'] == "Cancelled":
                cancelled_count += 1
                
        for route in routes:
            if route['available_seats'] == 0:
                fully_booked_count += 1
                
        print("\nBOOKING REPORT")
        print(f"Total bookings created: {total_created}")
        print(f"Confirmed bookings: {confirmed_count}")
        print(f"Cancelled bookings: {cancelled_count}")
        print(f"Seats currently reserved: {reserved_seats}")
        print(f"Confirmed revenue: Rs. {confirmed_revenue}")
        print(f"Fully booked routes: {fully_booked_count}")

    elif choice == 8:
        print("\nROUTE SCHEDULE BY DEPARTURE CITY")
        for city in supported_cities:
            print(f"\n{city}")
            has_route = False
            for route in routes:
                if route['departure_city'] == city:
                    print(f"  {route['route_id']} -> {route['destination_city']} | {route['departure_time']}")
                    has_route = True
            if not has_route:
                print("  No scheduled departures.")
    elif choice == 9:
        print("Thanks for using this application")
        break
    else:
        print("Something is wrong")
```
