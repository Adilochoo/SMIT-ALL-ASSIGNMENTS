```python
all_accounts=[]
```

```python
print('''
____________________________________________________________________________________________

                            Mezan Bank Limited
____________________________________________________________________________________________



        1 Prees : Open New Account

        2 Press : View Your Account

        3 Press : Deposit

        4 Press : Withdraw

        5 Press : Quit
''')
for i in range(200):
    choice=int(input('enter your choice : '))
    if choice==1:
        name=input('enter your name : ')
        age=int(input('enter the age :'))
        gender=input('enter the gender : ')
        initial_deposit=int(input('enter the initial deposit amount in pkr : '))
        account_number=random.randint(3333,9999)
        all_accounts.append({'account_number':account_number,'Name':name,
                             'age':age,'gender':gender,'balance':initial_deposit})
        print(f'Mr/Mrs {name} your account number is {account_number} your account has been created sucessfully')
    elif choice==2:
        account_number=int(input('enter your account number : '))
        for account in all_accounts:
            if account['account_number']==account_number:
                print(f'''
__________________________________________________________________________________________

                            Mezan Bank Limited 
__________________________________________________________________________________________


       Account Number : {account['account_number']}           Name : {account['Name']}
       
       Age : {account['age']} years old                    Gender : {account['gender']}
       
    Balance : {account['balance']}''')
                break
        else:
            print('Account not found')
    elif choice==3:
        account_number=int(input('enter the account number : '))
       
        for account in all_accounts:
            if account['account_number']==account_number:
                amount=int(input('enter the deposit amount'))
                account['balance']+=amount
                print(f'Mr/Mrs {account['Name']} your deposit Rs : {amount} are deposit sucessfully')
                break
        else:
            print('Invalid account')
    elif choice==4:
        account_number=int(input('enter the account number : '))
        for account in all_accounts:
            if account['account_number']==account_number:
                amount=int(input('enter the withdraw amount'))
                if amount<=account['balance']:
                    account['balance']-=amount
                    print(f'Mr/Mrs {account['Name']} your withdraw Rs : {amount} are withdraw sucessfully')
                    break
                else:
                    print('Insufficant Balance')
                    break
        
        else:
            print('Invalid account')
    elif choice==5:
        print(f'Thanks for using this Mezan App')
        break
    else:
        print('Cammand Not found')
        
```
